## A. Assess Baseline Model

```python
# DO NOT MODIFY THE CODE IN THIS CELL
# Load data
try:
  X_train = pd.read_csv(folder_path / 'X_train.csv')
  y_train = pd.read_csv(folder_path / 'y_train.csv')

  X_val = pd.read_csv(folder_path / 'X_val.csv')
  y_val = pd.read_csv(folder_path / 'y_val.csv')

  X_test = pd.read_csv(folder_path / 'X_test.csv')
  y_test = pd.read_csv(folder_path / 'y_test.csv')
except Exception as e:
  print(e)
```

### A.1 Generate Predictions with Baseline Model

```python
base_reg = DummyRegressor(strategy='mean')
base_reg.fit(X_train, y_train)   ## Finding the constant or mean of target variable (net_premium_amount)
```

```python
# Generate predictions
y_preds_training = base_reg.predict(X_train)
y_preds_validation = base_reg.predict(X_val)
```

```python
### Printing the mean or constant value of the training set.
print(y_preds_training)
print('  ')
print(y_preds_validation)
```

### A.2 Selection of Performance Metrics

> In this project, I decided to use RMSE and MAE as model evaluation.


1.   MAE provide the average size of the error (between actual and predicted). In addition, it show the same unit of error toward the targer variable.
2.   RMSE is sensive to the outlier which penalise the model more largely than MAE.

Additionally, in this project we will focus on MAE which has the same unit as targer variable.



```python
# In this case, I decide to use RMSE and MAE as a performance metric.

from sklearn.metrics import root_mean_squared_error as rmse
from sklearn.metrics import mean_absolute_error as mae
```

```python
performance_metrics_explanations = """
MAE and RMSE provide the same unit of the target variable which MAE provide the average size of the error which it is convenience to interpret
"""
```

### A.3 Baseline Model Performance

> Provide some explanations on model performance


```python
## perform the baseline model for comparing to the modelling part
print("The performance of Training Baseline model is:")
print('RMSE of Training: ',rmse(y_train, y_preds_training))
print('MAE of Training: ',mae(y_train, y_preds_training))

print(' ')

print("The performance of Validation Baseline model is:")
print('RMSE of Validation: ', rmse(y_val, y_preds_validation))
print('MAE of Validation: ', mae(y_val, y_preds_validation))
```
