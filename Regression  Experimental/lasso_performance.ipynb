## C. Train Machine Learning Model

### C.1 Import Algorithm



```python
from sklearn.linear_model import Lasso

lasso_reg1 = Lasso()
lasso_reg1.fit(X_train[features_list], y_train)  ## For training set only, The algorithm learn from the data.
```

```python
## The prediction result comparison for the Lasso model when alpha = 1 for the testing set
# The result are in form of RMSE and MAE

y_preds_training = lasso_reg1.predict(X_train[features_list])

print("The performance of Lasso training model is:")
print('RMSE of Training: ', rmse(y_train, y_preds_training))
print('MAE of Training: ', mae(y_train, y_preds_training))

# Overall, Lasso model while alpha = 1 has a better performance than a baseline model (training set)
```

```python
y_preds_validation = lasso_reg1.predict(X_val[features_list])

print("The performance of Lasso Validation set is:")
print('RMSE of Validation: ', rmse(y_val, y_preds_validation))
print('MAE of Validation: ', mae(y_val, y_preds_validation))

## However, the model perform of validation set better than baseline model.
```
