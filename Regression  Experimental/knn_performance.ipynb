## C. Train Machine Learning Model

### C.1 Import Algorithm



```python
# Import the library for KNN model
from sklearn.neighbors import KNeighborsRegressor

from sklearn.metrics import root_mean_squared_error as rmse
from sklearn.metrics import mean_absolute_error as mae
```

```python
## Generating the Default result for KNN model
# this is not act like basedline model but this show the result of the KNN with the default hyperameter.

default_knn = KNeighborsRegressor()
default_knn.fit(X_train, y_train)   ## The let the algorithm learn from the testing data.

# Then, let it predict outcome from independent features
y_train_preds_default = default_knn.predict(X_train)  ## For training set
y_val_preds_default = default_knn.predict(X_val)


# Lastly, display the matrics result for both training and validation set.
# Training KNN result
print("The performance of KNN default training model is:")
print('RMSE of Training: ', rmse(y_train, y_train_preds_default))
print('MAE of Training: ', mae(y_train, y_train_preds_default))

print("  ")

# Validation result
print("The performance of KNN default on validation set:")
print('RMSE of Validation: ', rmse(y_val, y_val_preds_default))
print('MAE of Validation: ', mae(y_val, y_val_preds_default))
```
