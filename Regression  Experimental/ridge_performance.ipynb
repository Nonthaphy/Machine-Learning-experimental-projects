## C. Train Machine Learning Model

### C.1 Import Algorithm



```python
from sklearn.linear_model import Ridge

reg1 = Ridge()
reg1.fit(X_train[features_list], y_train)  ## For training set only, due to if we use .fit for validation set, it will retain the model by using validation set which causing bias.
```

```python
## The result of the Ridge model of the training set
y_preds_training = reg1.predict(X_train[features_list])

print("The performance of Ridge training model is:")
print('RMSE of Training: ', rmse(y_train, y_preds_training))
print('MAE of Training: ', mae(y_train, y_preds_training))

## As result, the model has a slightly better performance than the baseline model (MAE baseline = 37.75)
```

```python
## The result for the validation set when compared to the baseline.

y_preds_validation = reg1.predict(X_val[features_list])

print("The performance of Ridge Validation set is:")
print('RMSE of Validation: ', rmse(y_val, y_preds_validation))
print('MAE of Validation: ', mae(y_val, y_preds_validation))

## Also, the model perform of validation set slightly better than baseline model (MAE baseline = 137.54).
```
