## G. Train Machine Learning Model

### G.1 Import Algorithm


```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import RandomizedSearchCV, GridSearchCV, StratifiedKFold, KFold
```

```python
# Generating the prediction result without hyper-tuning (default)
et_default = ExtraTreesClassifier(random_state = 42)

# Fit the model on training set
et_default.fit(X_train, y_train)

# Making the prediction on the training and validation set.
y_train_pred_default = et_default.predict(X_train)
y_val_pred_default = et_default.predict(X_val)
```

```python
# Train default result:
print('TRAINING Extra tree performace when hyperparameter = default:')
print('Accuracy: ',accuracy_score(y_train, y_train_pred_default))
print('F1-macro: ',f1_score(y_train, y_train_pred_default, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_train_pred_default, average = 'weighted'))
print("\n" *1)

# Valdidation default result:
print('VALIDATION Extra tree performace when hyperparameter = default:')
print('Accuracy: ',accuracy_score(y_val, y_val_pred_default))
print('F1-macro: ',f1_score(y_val, y_val_pred_default, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_val_pred_default, average = 'weighted'))

# From the result, the training set is overfitting at every metric = 1 (perfectly overfitting)
# Accuracy and F1-weighted has a similar score (almost the same indicating the model generalise well on validation set.)
```

```python
# <Student to fill this section>
algorithm_selection_explanations = """
An extra tree model is similar to the Random Forest model, which creates more trees, but it introduces randomness in the way splits are chosen, which reduces variance and leads to generally better performance on unseen data.
On the other hand, Extra Trees increase model diversity leading to model consistency.
Model with default parameters show the result that training set is perfectly overfitting and validation set has a very score with DT and RF model.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='algorithm_selection_explanations', value=algorithm_selection_explanations)
```

### G.2 Set Hyperparameters

```python
# Define the hyperparameter that we are going to use in Extra tree model
# In this model, we will increase the number of hyperparameters, which I believe will help to make this model more robust.
et_hyper_grid = {
    'n_estimators': [200, 300, 400],  # We try to reduce the minimum number of tree becasue more randomness of the model
    'max_depth': [18, 24, 36],
    'min_samples_leaf': [36, 48],
    'max_features': ['sqrt', 'log2'],
    'class_weight': [None, 'balanced']
}
```

```python
# <Student to fill this section>
hyperparameters_selection_explanations = """
Accoring to the main characteristic of this model is randomess, I considered that applying GridSearchCV is robust and increase a chance to find the best parameter from evey possiblie of parameter family.
In this parameter grid, we decrease the lowest number of n_estimators and min_samples_leaf to ensure that the model are capable to generalise on the unseen data (because this model introduce more randomess) .
In addition, the maximum number of each parameter is still the same because I want to know how it will improve from RF and StratifiedKFold with fold = 5 is still applied to ensure the class proportion of each fold.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='hyperparameters_selection_explanations', value=hyperparameters_selection_explanations)
```

### G.3 Fit Model

```python
# Define the model
et_model = ExtraTreesClassifier(random_state = 42, n_jobs=-1)

# Introducing StratifiedKFold for helping in generalising and training model
sf_kfold = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 42)

# Applying GridSearchCV to find the best hyperparameter
# According to the characteristics of the extra tree model, which rely on randomness.
# We decide to use GridSearchCV to ensure that we will get the best parameter.
grid_search_et = GridSearchCV(et_model, et_hyper_grid, cv = sf_kfold, verbose = 1)

# fit the training set.
grid_search_et.fit(X_train, y_train)
```

```python
# Display the best parameter
grid_search_et.best_params_

# Then save the parameter for refitting the model with our best parameter.
et_best_hyper = grid_search_et.best_estimator_
```

```python
et_best_hyper
```

### G.4 Model Technical Performance

```python
# Apply the best parameter to each set by using .predict()
# Remark: GridSearchCV has refit = default, which is not required to fit the training set again

y_preds_training = et_best_hyper.predict(X_train)
y_preds_validation = et_best_hyper.predict(X_val)
y_preds_testing = et_best_hyper.predict(X_test)
```

```python
# Train set Extra tree result:
print('TRAINING Extra tree performace:')
print('Accuracy: ',accuracy_score(y_train, y_preds_training))
print('F1-macro: ',f1_score(y_train, y_preds_training, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_preds_training, average = 'weighted'))
print("\n" *1)

# Valdidation set Extra tree result:
print('VALIDATION Extra tree performace:')
print('Accuracy: ',accuracy_score(y_val, y_preds_validation))
print('F1-macro: ',f1_score(y_val, y_preds_validation, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_preds_validation, average = 'weighted'))
print("\n" *1)

# Test set Extra tree result:
print('TESTING Extra tree performance:')
print('Accuracy: ',accuracy_score(y_test, y_preds_testing))
print('F1-macro: ',f1_score(y_test, y_preds_testing, average = 'macro'))
print('F1-weighted: ',f1_score(y_test, y_preds_testing, average = 'weighted'))
```

```python
# <Student to fill this section>
model_performance_explanations = """
ExtraTrees shows a good generalisation on the train/val/test set because their performance are very similar. The small gap between F1-macro and F1-weighted indicates it generalizes well across all classes.
When comparing val performance with the default model, the reason is similar.
However, The test performance of ExtraTree is a bit below RF and DT model while the different between F1-macro and F1-weighted is only 0.01 (RF and DT are different around 0.03)
"""
```
