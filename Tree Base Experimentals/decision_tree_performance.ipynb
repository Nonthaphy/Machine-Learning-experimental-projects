## G. Train Machine Learning Model

 > In our first experiment, we will apply a decision tree algorithm to predict credit_rating.

 > On model training, We will apply the stratified cross validation to ensure that each fold have a similar class proportions.

 > For searching CV, we considered to use GRID search CV to find the best hyperparameter which will find the best parameters from every possibility from the hyperparameter grid.

### G.1 Import Algorithm


```python
# Importing additional library
from sklearn.metrics import accuracy_score, f1_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import RandomizedSearchCV, GridSearchCV, StratifiedKFold, KFold
```

```python
# Firstly, we will train the model in prediction on the train and validation with the Default Hyperparameters
dt_0 = DecisionTreeClassifier(random_state = 42)

# Fit the model on train set
dt_0.fit(X_train, y_train)

# Making the prediction on the train and validation set
y_train_preds0 = dt_0.predict(X_train)
y_val_preds0 = dt_0.predict(X_val)
```

```python
# Train default result:
print('TRAINING Decision tree performace when hyperparameter = default:')
print('Accuracy: ',accuracy_score(y_train, y_train_preds0))
print('F1-macro: ',f1_score(y_train, y_train_preds0, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_train_preds0, average = 'weighted'))
print("\n" *1)


# Valdidation default result:
print('VALIDATION Decision tree performace when hyperparameter = default:')
print('Accuracy: ',accuracy_score(y_val, y_val_preds0))
print('F1-macro: ',f1_score(y_val, y_val_preds0, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_val_preds0, average = 'weighted'))

# Very interesting, the training set with the default hyperparameter is totally overfitting.
# However, on validation set, has a similar performance.
```

```python
# <Student to fill this section>
algorithm_selection_explanations = """
A decision tree is a very considerable model that creates only 1 tree and includes all the features that we have. I believe that the model will improve from the baseline performance.
The model with default parameters indicates that this model is perfectly overfitting due to a perfect score on the train set. However, validation set performance is improved especially on F1-score.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='algorithm_selection_explanations', value=algorithm_selection_explanations)
```

### G.2 Set Hyperparameters

```python
# Setting the set of hyperparameters into param_grid
dt_hyper_grid = {
    "max_depth": [3, 5, 7, 9, 12, 15, 20],
    "min_samples_leaf": [1, 5, 10, 20, 40, 50],
    "min_samples_split": [2, 10, 20, 50, 75, 100],
    "class_weight": [None, "balanced"]    ## Balance will make a algorithm consider minority class weight.
}
```

```python
# <Student to fill this section>
hyperparameters_selection_explanations = """
The set of parameters on the grid is designed for testing the wide range of possible outcomes.
I want to test the combination of smaller to larger numbers in the grid to find the most robust parameter that can overcome model overfitting.
Instead of find testing 1 by 1, I decided to apply StratifiedKFold and Gridsearch to ensure that we will find the best one.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='hyperparameters_selection_explanations', value=hyperparameters_selection_explanations)
```

### G.3 Fit Model

```python
# include the hyperparameter in the model
dt_model = DecisionTreeClassifier(random_state = 42)

# Applying StratifiedKFold to ensure the propotion to be similar across the fold to prevent model overfitting.
sf_kfold = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 42)   # We are splitting the train set into 5 folds and shuffling or mixing the rows.

# Applying Grid search to find the best hyperparameter from all possible parameter family on Training set
grid_search_dt = GridSearchCV(dt_model, dt_hyper_grid, cv = sf_kfold, verbose = 1)

# Then fit the training model
grid_search_dt.fit(X_train, y_train)
```

```python
# Display the best parameter
grid_search_dt.best_params_

# Then save the parameter for refitting the model with our best parameter.
dt_best_hyper = grid_search_dt.best_estimator_
```

```python
dt_best_hyper
```

### G.4 Model Technical Performance

```python
# Apply the best parameter to each set.
# Remark: GridSearchCV has refit = default, which is not required to fit the training set again

y_preds_training = dt_best_hyper.predict(X_train)
y_preds_validation = dt_best_hyper.predict(X_val)
y_preds_testing = dt_best_hyper.predict(X_test)
```

```python
# Train set Decision tree result:
print('TRAINING DECISION TREE performace:')
print('Accuracy: ',accuracy_score(y_train, y_preds_training))
print('F1-macro: ',f1_score(y_train, y_preds_training, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_preds_training, average = 'weighted'))
print("\n" *1)

# Valdidation set Decision tree result:
print('VALIDATION DECISION TREE performace:')
print('Accuracy: ',accuracy_score(y_val, y_preds_validation))
print('F1-macro: ',f1_score(y_val, y_preds_validation, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_preds_validation, average = 'weighted'))
print("\n" *1)

# Test set Decision tree result:
print('TESTING DECISION TREE performance:')
print('Accuracy: ',accuracy_score(y_test, y_preds_testing))
print('F1-macro: ',f1_score(y_test, y_preds_testing, average = 'macro'))
print('F1-weighted: ',f1_score(y_test, y_preds_testing, average = 'weighted'))
```

```python
# Comparing to the baseline model.
print('TRAINING BASELINE performace:')
print('Accuracy:  0.5397752808988764')
print('F1-macro:  0.2337030550690796')
print('F1-weighted:  0.37844139659051407')
print("\n" *1)

print('VALIDATION BASELINE performace:')
print('Accuracy:  0.5395327942497754')
print('F1-macro:  0.2336348604221379')
print('F1-weighted:  0.378161007233137')
```

```python
# <Student to fill this section>
model_performance_explanations = """
Overall, the model is considered minimal overfitting which training set is slightly better than the val and test sets, but it is acceptable.
Compared to the default model, all metrics of the validation set performance increased approximately 10% and the training set is not perfectly overfitting.
Furthermore, val and test set have a very similar performace indicating the model in generalise well on unseen data which match with our object (finding the model which consistency).
Interestingly, the F1-macro and F1-weighted a slightly similar (0.03 different), indicating the model generalizes well on every class.
"""
```
