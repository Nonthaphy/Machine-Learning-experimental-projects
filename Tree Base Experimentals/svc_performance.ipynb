## G. Train Machine Learning Model

### G.1 Import Algorithm


```python
# <Student to fill this section>
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import RandomizedSearchCV, GridSearchCV, StratifiedKFold, KFold
```

```python
# Creating an SVC model with the default hyperparameter
svc_default = SVC()

# Fit the model on training set
svc_default.fit(X_train, y_train)

# Making the prediction on the training and validation set.
y_train_pred_default = svc_default.predict(X_train)
y_val_pred_default = svc_default.predict(X_val)
```

```python
# Train default result:
print('TRAINING SVC performace when hyperparameter = default:')
print('Accuracy: ',accuracy_score(y_train, y_train_pred_default))
print('F1-macro: ',f1_score(y_train, y_train_pred_default, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_train_pred_default, average = 'weighted'))
print("\n" *1)

# Valdidation default result:
print('VALIDATION SVC performace when hyperparameter = default:')
print('Accuracy: ',accuracy_score(y_val, y_val_pred_default))
print('F1-macro: ',f1_score(y_val, y_val_pred_default, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_val_pred_default, average = 'weighted'))

# The training performance is slightly overfitting. (acceptable but requires regularisation)
# However, when we compare with Decision tree, Random forest and extra tree. All of their training (default parameter) is perfectly overfitting (= 1)
```

```python
# <Student to fill this section>
algorithm_selection_explanations = """
Instead of splitting the leaf, SVC draws smooth (non-linear) decision boundaries between class. In addition, it is easy to tweak the regularisation (not too complex like RF and DT) which might reduce the model noise.
According to default parameter, training set is a bit overfitting (but DT/RF/ET is perfectly overfitting on default train set) and validation performance is similar to other 3 models.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='algorithm_selection_explanations', value=algorithm_selection_explanations)
```

### G.2 Set Hyperparameters

```python
# Define the SVC hyperparameter grid.
svc_hyper_grid = {
    'C': [0.1, 1, 3, 10, 30],
    'gamma': ['scale', 'auto', 0.1, 0.01],
    'kernel': ['poly', 'rbf', 'sigmoid'],
    'degree': [2, 3, 4],
    'class_weight': [None, 'balanced']
}
```

```python
# <Student to fill this section>
hyperparameters_selection_explanations = """
In this parameter grid, I want to test and find the best commbination by using RandomizedSearchCV (faster and reduce computational cost).
Inside the grid, I set a wide range of C and gamma to find out how it is impact to the performance. 3 types of kernel is define to make a diversity to the model.
In summary, this parameter might increase the model performance.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='hyperparameters_selection_explanations', value=hyperparameters_selection_explanations)
```

### G.3 Fit Model

```python
# Define the model and hyperparameter grid.
svc_model = SVC()

# Apply StratifiedKFold
sf_kfold = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 42)

# In this case, we are going to use RandomizedSearchCV to find the most suitable parameter with the lowest computational cost
random_search_svc = RandomizedSearchCV(svc_model, svc_hyper_grid, cv = sf_kfold, n_jobs=-1, verbose = 1)

# fit the training set.
random_search_svc.fit(X_train, y_train)
```

```python
# Display the best parameter
random_search_svc.best_params_

# Then save the parameter for refitting the model with our best parameter.
svc_best_hyper = random_search_svc.best_estimator_
```

```python
svc_best_hyper
```

### G.4 Model Technical Performance

```python
# Apply the best parameter to each set by using .predict()

y_preds_training = svc_best_hyper.predict(X_train)
y_preds_validation = svc_best_hyper.predict(X_val)
y_preds_testing = svc_best_hyper.predict(X_test)
```

```python
# Train set SVC result:
print('TRAINING SVC performace:')
print('Accuracy: ',accuracy_score(y_train, y_preds_training))
print('F1-macro: ',f1_score(y_train, y_preds_training, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_preds_training, average = 'weighted'))
print("\n" *1)

# Valdidation set SVC result:
print('VALIDATION SVC performace:')
print('Accuracy: ',accuracy_score(y_val, y_preds_validation))
print('F1-macro: ',f1_score(y_val, y_preds_validation, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_preds_validation, average = 'weighted'))
print("\n" *1)

# Test set SVC result:
print('TESTING SVC performance:')
print('Accuracy: ',accuracy_score(y_test, y_preds_testing))
print('F1-macro: ',f1_score(y_test, y_preds_testing, average = 'macro'))
print('F1-weighted: ',f1_score(y_test, y_preds_testing, average = 'weighted'))
```

```python
# Comparing to the baseline model.
print('TRAINING BASELINE performace:')
print('Accuracy:  0.5397752808988764')
print('F1-macro:  0.2337030550690796')
print('F1-weighted:  0.37844139659051407')
print("\n" *1)

print('VALIDATION BASELINE performace:')
print('Accuracy:  0.5395327942497754')
print('F1-macro:  0.2336348604221379')
print('F1-weighted:  0.378161007233137')
```
