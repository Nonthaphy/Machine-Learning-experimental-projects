## G. Train Machine Learning Model

### G.1 Import Algorithm


```python
# <Student to fill this section>
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import RandomizedSearchCV, GridSearchCV, StratifiedKFold, KFold
```

```python
# Firstly, generating the result with the Random Forest default hyperparameter
rf_default = RandomForestClassifier(random_state = 42)

# Fit the model on training set
rf_default.fit(X_train, y_train)

# Making the prediction on the training and validation set.
y_train_pred_default = rf_default.predict(X_train)
y_val_pred_default = rf_default.predict(X_val)
```

```python
# Train default result:
print('TRAINING Random Forest performace when hyperparameter = default:')
print('Accuracy: ',accuracy_score(y_train, y_train_pred_default))
print('F1-macro: ',f1_score(y_train, y_train_pred_default, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_train_pred_default, average = 'weighted'))
print("\n" *1)

# Valdidation default result:
print('VALIDATION Random Forest performace when hyperparameter = default:')
print('Accuracy: ',accuracy_score(y_val, y_val_pred_default))
print('F1-macro: ',f1_score(y_val, y_val_pred_default, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_val_pred_default, average = 'weighted'))

# training set performacne is perfectly overfitting which all performance matrics = 1.0
# However, the validation performance is significantly higher than Decision Tree and Random Forest.
# Interestingly, validation set, has similar performance.
```

```python
# <Student to fill this section>
algorithm_selection_explanations = """
From my perspective, the Random Forest model is more robust than decision tree by creating more tree and applying bootstraping which help the model generalse better on unseen data.
On the the hand, the model will be trained on different combination of parameters on many tree that would outperform DT model.
The result of the model with default parameters shows that the model is still perfectly overfitting on default paramter (score = 1).
However, the val set performance increases from the DT model by around 10%.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='algorithm_selection_explanations', value=algorithm_selection_explanations)
```

### G.2 Set Hyperparameters

```python
# Setting the set of hyperparameters into param_grid
# According to many trials, we had selected a huge amount of parameter grid, and it took more than an hour to finish executing.
# on top of that, there are many time that the run time is disconnect. Then I decide to reduce the complexity of the parameters.

rf_hyper_grid = {
    'n_estimators': [100, 200, 300, 400],
    'max_depth': [18, 24, 36],
    'min_samples_leaf': [24, 36, 48],
    'max_features': ['sqrt', 'log2'],    # due to we have lots of feature (60 in total), using sqrt and log2 is a best choice and reducing the computational cost.
    'class_weight': [None, 'balanced']   # Balance will make a algorithm consider minority class weight in case of when the set or fold is imbalanec
}
```

```python
# <Student to fill this section>
hyperparameters_selection_explanations = """
In this experiment, we will introduce RandomiseSearchCV to find the best parameter due to It help in reducing executing time on searching.
Inside the grid, we have set a medium to large value to testing the combination and ensure that the model reflect the real world data.
'sqrt' and 'log2' will be use to define the max_features because we have 60 features in total and It must take many hours to execute which I considered not worth trying.
Also StratifiedKFold with fold = 5 is still ,applied to ensure the class proportion of each fold.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='hyperparameters_selection_explanations', value=hyperparameters_selection_explanations)
```

### G.3 Fit Model

In this model, we will use Randomise search CV to find the best parameter.
The reason is It take more than hours to run the model and I consider that it is not worth for the computational cost and reduce for making another model.

```python
# Next step, introduce a hyperparameter grid in the model
rf_model = RandomForestClassifier(random_state = 42, n_jobs=-1)

# Using StratifiedKFold to maintain the proportion. From my perspective, this will help the model generalize to the unseen data.
sf_kfold = StratifiedKFold(n_splits = 5, shuffle = True, random_state = 42)

# Introduce Randomise search to find the best and robust hyperparameter formula and apply kfold from Stratified.
random_search_rf = RandomizedSearchCV(rf_model, rf_hyper_grid, cv = sf_kfold, verbose = 1)

# fit the training set.
random_search_rf.fit(X_train, y_train)
```

```python
# Display the best parameter
random_search_rf.best_params_

# Then save the parameter for refitting the model with our best parameter.
rf_best_hyper = random_search_rf.best_estimator_
```

```python
rf_best_hyper
```

### G.4 Model Technical Performance

```python
# Apply the best parameter to each set by using .predict()
# Remark: RandomiseCV also refit the train set. It is mean, there is no need to use .fit() on train set again

y_preds_training = rf_best_hyper.predict(X_train)
y_preds_validation = rf_best_hyper.predict(X_val)
y_preds_testing = rf_best_hyper.predict(X_test)
```

```python
# Train set Random Forest result:
print('TRAINING Random forest performace:')
print('Accuracy: ',accuracy_score(y_train, y_preds_training))
print('F1-macro: ',f1_score(y_train, y_preds_training, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_preds_training, average = 'weighted'))
print("\n" *1)

# Valdidation set Random Forest result:
print('VALIDATION Random forest performace:')
print('Accuracy: ',accuracy_score(y_val, y_preds_validation))
print('F1-macro: ',f1_score(y_val, y_preds_validation, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_preds_validation, average = 'weighted'))
print("\n" *1)

# Test set Random Forest result:
print('TESTING Random forest performance:')
print('Accuracy: ',accuracy_score(y_test, y_preds_testing))
print('F1-macro: ',f1_score(y_test, y_preds_testing, average = 'macro'))
print('F1-weighted: ',f1_score(y_test, y_preds_testing, average = 'weighted'))
```

```python
# Comparing to the baseline model.
print('TRAINING BASELINE performace:')
print('Accuracy:  0.5397752808988764')
print('F1-macro:  0.2337030550690796')
print('F1-weighted:  0.37844139659051407')
print("\n" *1)

print('VALIDATION BASELINE performace:')
print('Accuracy:  0.5395327942497754')
print('F1-macro:  0.2336348604221379')
print('F1-weighted:  0.378161007233137')
```
