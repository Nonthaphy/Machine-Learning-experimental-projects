### A.1 Generate Predictions with Baseline Model

```python
# Baseline model for our classification model; Decision tree, Random Forest, Extra tree and SVC
from sklearn.dummy import DummyClassifier

# Creating a baseline model
classi_baseline = DummyClassifier(strategy = 'most_frequent')   ## strategy = 'most_frequent' is we are going to predict the majority class of
classi_baseline.fit(X_train, y_train)
```

### A.2 Selection of Performance Metrics

> In this project 2 metrics will be use to evalute model performance

* F1 score: It evaluates the model's performance in predicting correctly. It mean, from our prediction, how many customer belong to each class.
* accuracy: indicating the model overall accuracy for the prediction.


```python
# Importing the accuracy and F1 score
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
```

```python
performance_metrics_explanations = """
In this project, my models will be evaluated by an accuracy score that reflects how many the model predicts correctly.
And f1_score assesses deeply which indicate from the prediction on the standard, good and poor class, how many that we predict correctly (capture the class level).
Furthermore, a higher F1 score indicates that the model predicts the right class with few mistakes or misses, or can capture lots of classes, which balances precision and recall.
To find the consistency model, I consider these 2 metrics to be the most suitable.
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='performance_metrics_explanations', value=performance_metrics_explanations)
```

### A.3 Baseline Model Performance

> Provide some explanations on model performance


```python
## Applying .predict() to train and  validation sets
# For training set:
y_preds_training = classi_baseline.predict(X_train)
y_preds_validation = classi_baseline.predict(X_val)

# Train Baseline result:
print('TRAINING BASELINE performace:')
print('Accuracy: ',accuracy_score(y_train, y_preds_training))
print('F1-macro: ',f1_score(y_train, y_preds_training, average = 'macro'))
print('F1-weighted: ',f1_score(y_train, y_preds_training, average = 'weighted'))
print("\n" *1)

# Valdidation Baseline result:
print('VALIDATION BASELINE performace:')
print('Accuracy: ',accuracy_score(y_val, y_preds_validation))
print('F1-macro: ',f1_score(y_val, y_preds_validation, average = 'macro'))
print('F1-weighted: ',f1_score(y_val, y_preds_validation, average = 'weighted'))
```

```python
baseline_performance_explanations = """
Overall, the results of the train and val baseline have a similar dataset. They got an accuracy of 0.539, indicating 54% of their predictions are correct (seems a bit low in my perspective).
Macro-F1 is used for treating all class equally and they got 0.23 which is very low indicating the model might weight 1 class over another.
F1-weighted is use in giving the weight on the majority class. they have 0.378 that is considered low.
In summary, their score are almost the same, indicating the model is stable but poor performance.
However, these are just baseline performances that we will use them to compare to algorithm performance.
"""
```
